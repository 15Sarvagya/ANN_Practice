Here we have generated a model but still we have confusion about whether our optimizer is good enough. Should I change it to another and try again and again? Are the number of my hidden layers enough, are the number of my neurons enough, and are many other parameters enough to overcome this problem? To overcome this problem we have Keras Tuner, which tunes the hyperparameters and tells us which parameter will fit much better (such as the number of neurons, the number of hidden layers, the optimizer, etc.).

Here we are going to do it in different orders:
1. How to select the output up to `max`
2. Number of nodes in a layer
3. How to select the number of layers
4. All in one model

1]Hyper-tuning optimizers only for now.

```python
import keras_tuner as kt
```

```python
def build_model(hp):
  model=Sequential()
  model.add(Input(shape=(8,)))
  model.add(Dense(32,activation="relu"))
  model.add(Dense(1,activation="sigmoid"))
  optimizer = hp.Choice("optimizer",values=["adam","sgd","rsmprop","adadelta"])
  model.compile(optimizer=optimizer,loss="binary_crossentropy",metrics=["accuracy"])
  return model
```

```python
tuner=kt.RandomSearch(build_model,max_trials=5,objective="val_accuracy")
```

```python
tuner.search(x_train,y_train,epochs=7,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters(num_trials=1)[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,epochs=100,initial_epoch=8,validation_data=(x_test,y_test))
```

See getting a much better result after hyperparameter tuning.

2] we are doing it for the number of nodes my hidden layer should have.


```python
def build_models(hp):
  model=Sequential()
  units=hp.Int("units",min_value=8,max_value=128,step=8)
  model.add(Input(shape=(8,)))
  model.add(Dense(units=units,activation="relu"))
  model.add(Dense(1,activation="sigmoid"))
  model.compile(optimizer="Adam",loss="binary_crossentropy",metrics=["accuracy"])
  return model
```

```python
tuner = kt.RandomSearch(build_models,max_trials=5,objective="val_accuracy",directory="mydir")
```

```python
tuner.search(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(x_train,y_train,epochs=100,initial_epoch=6,batch_size=32,validation_data=(x_test,y_test))
```

3], how to select the number of layers?

```python
def build_model(hp):
  model=Sequential()
  model.add(Input(shape=(8,)))
  num_layers = hp.Int(
    "num_layers",
    min_value=1,
    max_value=10
)
  for i in range(num_layers):
    model.add(Dense(72,activation="relu"))
  model.add(Dense(1,activation="sigmoid"))
  model.compile(loss="binary_crossentropy",optimizer="Adam",metrics=["accuracy"])
  return model
```

```python
tuner = kt.RandomSearch(build_model,objective="val_accuracy",max_trials=5, directory="mydir",
    project_name="num_layers_new",
    overwrite=True)
```

```python
tuner.search(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(x_train,y_train,validation_data=(x_test,y_test),epochs=100,initial_epoch=6)
```
