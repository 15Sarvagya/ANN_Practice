```python
class GDRegressor():
  def __init__(self,learning_rate,epochs):
    self.m = 33.56910203
    self.b= -120
    self.lr=learning_rate
    self.epochs= epochs
  def fit(self,x,y):
    for i in range(self.epochs):
      last_slope= -2*np.sum(y- self.m*x.ravel()- self.b)
      step_size = self.lr*last_slope
      self.b = self.b- step_size
    print(self.b)



```

```python
gd=GDRegressor(0.001,100)
```

```python
gd.fit(x,y)
```
