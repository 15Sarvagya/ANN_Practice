We have observed that, while doing a classification problem, not having dropout and having `n` number of epochs with heavy neurons, we are not gaining the accuracy that we have got on the training dataset. This is because of overfitting.

In the last graph the accuracy on the training set is so high and the accuracy on the test set is so low. The same goes for the loss: loss on the test set is so high and loss on training is very low. That's because of overfitting. To handle this we are going to use the dropout methods.
