```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.optimizers import Adam
```

```python
model= Sequential([
    Input(shape=(2,)),
    Dense(128,activation="linear"),
    Dense(128,activation="linear") ,
    Dense(1,activation="sigmoid")
])
```

```python
adam = Adam(learning_rate=0.01)
model.compile(loss = "binary_crossentropy",
optimizer = "adam",
metrics = ["accuracy"])
```

```python
history=model.fit(x,y,epochs=500,validation_split=0.2,verbose=1)
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x, y.astype('int'), clf=model, legend=2)
plt.xlim(-0.7,0.5)
plt.ylim(-0.8,0.8)
plt.title('p = 0.2')
plt.show()
```

See when we are talking about an activation function if we are not providing any activation function we suppose that the activation function is linear so it performs like a regression problem. It will not give me the classification solution for it. That's why we are not using a linear activation function for a classification problem because it will be giving me regression solutions. That's why we are using artificial activation functions like ReLU, sigmoid, or softmax to give me the classification solutions.

In short I just want to say that if we do not have an activation function the output of the classification problem will not be feasible. It will give me only regression solutions. For the classification solution we are using an activation function.
