```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y= make_regression(n_samples=4,n_features=1,noise=80,n_targets=1,n_informative=1,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(x,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,model.predict(x),c="r")
```

```python
#let m= 63.08504508
#b =0
y_pred=((78.35063668*x)+0).reshape(4)
y_pred
```

```python
plt.scatter(x,y)
plt.plot(x,model.predict(x),c='r',label='ols')
plt.plot(x,y_pred,label='b=0')
plt.legend()
plt.show()
```
