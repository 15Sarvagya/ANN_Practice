```python
model.get_weights()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history = model.fit(x,y,epochs=100,validation_split=0.2)
```

```python
model = Sequential()

model.add(Dense(10,activation='sigmoid',input_dim=2))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.get_weights()
```

```python
initial_weights = model.get_weights()
```

```python
initial_weights[0] = np.zeros(model.get_weights()[0].shape)
initial_weights[1] = np.zeros(model.get_weights()[1].shape)
initial_weights[2] = np.zeros(model.get_weights()[2].shape)
initial_weights[3] = np.zeros(model.get_weights()[3].shape)
```

```python
model.set_weights(initial_weights)
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
history = model.fit(x,y,epochs=100,validation_split=0.2)
```

```python
model.get_weights()
```

See we observe that in every layer the weights and the bias values are the same. That's because of the sigmoid activation function, which I already wrote about in the notebook. That's what the sigmoid function does.

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x,y.astype('int'), clf=model, legend=2)
```

See this is acting as a linear model. It is not differentiating between two classes or two outputs. You can see the bottom line in the graphs.
